```python
import polars as pl
from pathlib import Path
```

```python


print("Notebook folder:", Path.cwd())
RAW_DIR = Path.cwd().parent.parent / "data" / "raw" / "green_taxi"
print("Data folder:   ", RAW_DIR, "| exists:", RAW_DIR.exists())
print(list(RAW_DIR.glob("*.parquet")))
```

```python
print(sorted(p.name for p in RAW_DIR.glob("*.parquet")))
```

```python
df = pl.read_parquet(RAW_DIR / "green_tripdata_2025-01.parquet")
df.head()
print(df.shape)       # rows, columns
print(df.schema)      # column names and types

```

```python
for f in sorted(RAW_DIR.glob("*.parquet")):
    print(f.name, pl.read_parquet_schema(f))
```

```python
df.describe()                       # min/max, nulls for every column
```

```python
# Trips outside January 2025?
df.filter(
    (pl.col("lpep_pickup_datetime").dt.year() != 2025)
    | (pl.col("lpep_pickup_datetime").dt.month() != 1)
).height
```

```python
# Dropoff before pickup?
df.filter(pl.col("lpep_dropoff_datetime") < pl.col("lpep_pickup_datetime")).height
```

```python
# Exact duplicate rows?
df.height - df.unique().height
```

```python
# Nulls per column
df.null_count()
```
