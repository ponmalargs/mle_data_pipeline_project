```python
from pathlib import Path
import requests
import pandas as pd

BASE_URL = "https://d37ci6vzurychx.cloudfront.net/trip-data"
```

```python
# Folder containing this script -> its parent -> data/raw/green_taxi
#PROJECT_ROOT = Path(__file__).resolve().parent.parent
PROJECT_ROOT = Path.cwd().parent
RAW_DIR = PROJECT_ROOT / "data" / "raw" / "green_taxi"
RAW_DIR.mkdir(parents=True, exist_ok=True)
```

```python
YEAR = 2025
MONTHS = [1, 2, 3]
```

```python
def download(url: str, dest: Path) -> None:
    if dest.exists():
        print(f"Skip (exists): {dest.name}")
        return
    tmp = dest.with_suffix(".parquet.part")
    with requests.get(url, stream=True, timeout=120) as r:
        r.raise_for_status()
        with open(tmp, "wb") as f:
            for chunk in r.iter_content(chunk_size=1024 * 1024):
                f.write(chunk)
    tmp.rename(dest)
    print(f"Downloaded: {dest.name}")
```

```python
for month in MONTHS:
    url = f"{BASE_URL}/green_tripdata_{YEAR}-{month:02d}.parquet"
    dest = RAW_DIR / f"green_tripdata_{YEAR}-{month:02d}.parquet"
    download(url, dest)
```
